```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 105, 106, 107],
    "Name": ["usha", "Ravi", "PRIYA", "Kiran", "Sita", "Sita", "Arun", "Meena"],
    "Gender": ["F", "Male", "female", "M", "Female", "Female", "male", "F"],
    "Age": [22, 25, np.nan, 28, 30, 30, 150, 24],
    "Salary": [25000, 30000, 28000, np.nan, 35000, 35000, 40000, 27000],
    "Join_Date": ["2024-01-10", "10/02/2024", "2024-03-15",
                  "15-04-2024", "2024-05-20", "2024-05-20",
                  "2024-06-10", "2024-07-01"]
}

df = pd.DataFrame(data)

df
```

```python
# Display dataset information
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Check missing values in each column
print("Missing Values:")
print(df.isnull().sum())
```

```python
# Count duplicate rows
print("Number of Duplicate Rows:", df.duplicated().sum())
```

```python
# Data quality report
quality_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": df.isnull().sum().values,
    "Duplicate Values": [df[col].duplicated().sum() for col in df.columns],
    "Data Type": df.dtypes.astype(str).values
})

display(quality_report)
```

```python
# Check minimum and maximum values
print("Age Range:", df["Age"].min(), "to", df["Age"].max())

print("Salary Range:", df["Salary"].min(), "to", df["Salary"].max())

# Display records with unusual age values
display(df[df["Age"] > 100])
```

```python
# Fill missing Age with median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary with median
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Check missing values again
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Count duplicates before removal
print("Duplicates before removal:", df.duplicated().sum())

# Remove duplicate rows
df = df.drop_duplicates()

# Check duplicates after removal
print("Duplicates after removal:", df.duplicated().sum())

# Display updated dataset shape
print("Dataset Shape:", df.shape)
```

```python
# Standardise names
df["Name"] = df["Name"].str.title()

# Standardise gender values
df["Gender"] = df["Gender"].replace({
    "F": "Female",
    "female": "Female",
    "M": "Male",
    "male": "Male"
})

# Display updated values
print(df[["Name", "Gender"]])
```

```python
# Convert Join_Date into datetime format
df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    format="mixed",
    dayfirst=True,
    errors="coerce"
)

# Display updated data types
print(df.dtypes)

# Display dates
display(df[["Join_Date"]])
```

```python
# Replace unrealistic age values with the median
median_age = df.loc[df["Age"] <= 100, "Age"].median()

df.loc[df["Age"] > 100, "Age"] = median_age

# Check updated age values
print("Updated Age Values:")
print(df["Age"])

print("\nMaximum Age:", df["Age"].max())
```

```python
# Before and after cleaning summary
summary = pd.DataFrame({
    "Metric": [
        "Total Rows",
        "Total Columns",
        "Missing Values",
        "Duplicate Rows",
        "Maximum Age"
    ],
    "Before Cleaning": [
        8,
        6,
        2,
        1,
        150
    ],
    "After Cleaning": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["Age"].max()
    ]
})

display(summary)
```

```python
# Save cleaned dataset
df.to_csv("Cleaned_Employee_Data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
